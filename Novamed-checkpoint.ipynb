```python
import pandas as pd
import numpy as np
```

```python
# ─────────────────────────────────────────────────────────────
# LOAD RAW DATA
# ─────────────────────────────────────────────────────────────
 
df = pd.read_excel("C:/Users/Administrator/Downloads/NovaMed_Raw_Data.xlsx")
raw_count = len(df)
print(f"  Rows    : {df.shape[0]}")
print(f"  Columns : {df.shape[1]}")
print(f"  Columns : {list(df.columns)}")
```

```python
## INITIAL INSPECTION
# ─────────────────────────────────────────────────────────────
print("\n[STEP 2] Initial data inspection...")
 
print(f"\n  Data types:\n{df.dtypes}")
print(f"\n  Missing values per column:\n{df.isnull().sum()}")
print(f"\n  Total missing values : {df.isnull().sum().sum()}")
print(f"  Duplicate rows       : {df.duplicated().sum()}")
print(f"  Negative quantities  : {(df['quantity_sold'] < 0).sum()}")
print(f"  Zero revenue rows    : {(df['revenue_ngn'] == 0).sum()}")
print(f"\n  Sample of date column:\n{df['date'].head(10).tolist()}")
 
```

```python
## REMOVE DUPLICATE ROWS
# ─────────────────────────────────────────────────────────────
 
before = len(df)
df     = df.drop_duplicates().reset_index(drop=True)
after  = len(df)
 
print(f"  Rows before : {before}")
print(f"  Rows after  : {after}")
print(f"  Removed     : {before - after} duplicates")
```

```python
## STANDARDISE DATE COLUMN (MIXED FORMATS)
# ─────────────────────────────────────────────────────────────

def parse_date(val):
    """
    Try multiple date formats explicitly.
    """
    if pd.isnull(val):
        return pd.NaT

    val = str(val).strip()

    # Try each format in order
    formats = [
        "%Y-%m-%d",    # 2024-01-25  (ISO standard)
        "%d/%m/%Y",    # 25/01/2024  (DD/MM/YYYY)
        "%m/%d/%Y",    # 01/25/2024  (MM/DD/YYYY)
        "%d-%m-%Y",    # 25-01-2024  (DD-MM-YYYY)
        "%Y/%m/%d",    # 2024/01/25  (ISO with slashes)
        "%d %b %Y",    # 25 Jan 2024 (text month)
        "%B %d, %Y",   # January 25, 2024
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(val, format=fmt)
        except (ValueError, TypeError):
            continue

    # Last resort — let pandas infer
    try:
        return pd.to_datetime(val, infer_datetime_format=True)
    except:
        return pd.NaT

before = len(df)
df["date"] = df["date"].apply(parse_date)

nat_count = df["date"].isnull().sum()
print(f"  Total dates parsed    : {before - nat_count:,}")
print(f"  Still unparseable     : {nat_count}")
print(f"  % successfully parsed : {((before - nat_count) / before * 100):.1f}%")

# Only drop rows where date is STILL NaT after trying all formats
if nat_count > 0:
    print(f"  Dropping {nat_count} truly unparseable rows...")
    df = df.dropna(subset=["date"]).reset_index(drop=True)

print(f"  Rows remaining        : {len(df):,}")
print(f"  Date range            : {df['date'].min().date()} → {df['date'].max().date()}")
```

```python
# ─────────────────────────────────────────────────────────────
# STANDARDISE TEXT COLUMN FORMATTING
# ─────────────────────────────────────────────────────────────
 
# Applying strip + title case to all relevant string columns
text_cols = ["region", "channel", "category", "product_name", "state", "sales_rep"]
for col in text_cols:
    df[col] = df[col].str.strip().str.title()
 
# Verify unique values after standardisation
for col in ["region", "channel", "category"]:
    uniq = sorted(df[col].dropna().unique())
    print(f"  {col:<15} → {uniq}")
```

```python
# ─────────────────────────────────────────────────────────────
# HANDLE MISSING VALUES
# ─────────────────────────────────────────────────────────────
 
# missing sales_rep: cannot impute a person's name so mark as Unassigned
missing_rep = df["sales_rep"].isnull().sum()
df["sales_rep"] = df["sales_rep"].fillna("Unassigned")
print(f"  sales_rep    : {missing_rep} nulls → filled with 'Unassigned'")
 
# channel: fill with mode (most common channel)
missing_chan  = df["channel"].isnull().sum()
mode_channel  = df["channel"].mode()[0]
df["channel"] = df["channel"].fillna(mode_channel)
print(f"  channel      : {missing_chan} nulls → filled with mode '{mode_channel}'")
 
# discount_pct: assume 0% if no discount recorded
missing_disc      = df["discount_pct"].isnull().sum()
df["discount_pct"] = df["discount_pct"].fillna(0)
print(f"  discount_pct : {missing_disc} nulls → filled with 0")
 
print(f"\n  Total remaining nulls: {df.isnull().sum().sum()}")
```

```python
# ─────────────────────────────────────────────────────────────
# REMOVE INVALID TRANSACTIONS
# ─────────────────────────────────────────────────────────────
 
before = len(df)
 
# Negative quantity_sold: physically impossible which means they are data entry errors
neg_qty = (df["quantity_sold"] < 0).sum()
df      = df[df["quantity_sold"] > 0]
print(f"  Removed {neg_qty} rows with negative quantity_sold")
 
# Zero revenue: likely cancelled/voided orders with no analytical value
zero_rev = (df["revenue_ngn"] == 0).sum()
df       = df[df["revenue_ngn"] > 0].reset_index(drop=True)
print(f"  Removed {zero_rev} rows with zero revenue_ngn")
 
after = len(df)
print(f"\n  Rows before : {before}")
print(f"  Rows after  : {after}")
print(f"  Removed     : {before - after} invalid records")
```

```python
# ─────────────────────────────────────────────────────────────
# CORRECT DATA TYPES
# ─────────────────────────────────────────────────────────────
 
df["quantity_sold"]     = df["quantity_sold"].astype(int)
df["discount_pct"]      = df["discount_pct"].astype(float).astype(int)
df["unit_cost_ngn"]     = df["unit_cost_ngn"].astype(float)
df["unit_price_ngn"]    = df["unit_price_ngn"].astype(float)
df["revenue_ngn"]       = df["revenue_ngn"].round(2)
df["cogs_ngn"]          = df["cogs_ngn"].round(2)
df["gross_profit_ngn"]  = df["gross_profit_ngn"].round(2)
 
print(f"\n  Final data types:\n{df.dtypes}")
 
```

```python
# ─────────────────────────────────────────────────────────────
# ADD DERIVED COLUMNS
# ─────────────────────────────────────────────────────────────
 
# Time-based columns for trend analysis
df["date"]        = pd.to_datetime(df["date"])
df["year"]        = df["date"].dt.year
df["month"]       = df["date"].dt.month
df["month_name"]  = df["date"].dt.strftime("%B")
df["quarter"]     = df["date"].dt.quarter.map({1: "Q1", 2: "Q2", 3: "Q3", 4: "Q4"})
df["year_month"]  = df["date"].dt.strftime("%Y-%m")
 
# Profitability metric
df["gross_margin_pct"] = (
    (df["gross_profit_ngn"] / df["revenue_ngn"]) * 100
).round(2)
 
# Revenue segmentation
df["revenue_band"] = pd.cut(
    df["revenue_ngn"],
    bins   = [0, 50000, 200000, 500000, float("inf")],
    labels = ["Low (<₦50K)", "Medium (₦50K-200K)", "High (₦200K-500K)", "Very High (>₦500K)"]
)
 
print("  Columns added: year, month, month_name, quarter, year_month")
print("  Columns added: gross_margin_pct, revenue_band")
print(f"  Total columns now: {len(df.columns)}")
 
 
```

```python
# ─────────────────────────────────────────────────────────────
# FINAL VALIDATION
# ─────────────────────────────────────────────────────────────
 
print(f"\n  {'Metric':<30} {'Value'}")
print(f"  {'─'*50}")
print(f"  {'Raw rows':<30} {raw_count:,}")
print(f"  {'Clean rows':<30} {len(df):,}")
print(f"  {'Rows removed':<30} {raw_count - len(df):,} ({(raw_count - len(df)) / raw_count * 100:.1f}%)")
print(f"  {'Total columns':<30} {len(df.columns)}")
print(f"  {'Missing values':<30} {df.isnull().sum().sum()}")
print(f"  {'Duplicate rows':<30} {df.duplicated().sum()}")
print(f"  {'Negative quantities':<30} {(df['quantity_sold'] < 0).sum()}")
print(f"  {'Zero revenue rows':<30} {(df['revenue_ngn'] == 0).sum()}")
print(f"  {'Date range':<30} {df['date'].min().date()} → {df['date'].max().date()}")
print(f"  {'Unique products':<30} {df['product_name'].nunique()}")
print(f"  {'Unique regions':<30} {df['region'].nunique()}")
print(f"  {'Unique channels':<30} {df['channel'].nunique()}")
 
```

```python
# ─────────────────────────────────────────────────────────────
# SAVE CLEANED DATASET
# ─────────────────────────────────────────────────────────────
 
df.to_excel("C:/Users/Administrator/Downloads/NovaMed_Cleaned_data.xlsx", index=False)
df.to_csv("C:/Users/Administrator/Downloads/NovaMed_cleaned_data.csv",    index=False)
 
print("  Saved: Novamed_cleaned_data.xlsx")
print("  Saved: Novamed_cleaned_data.csv")
print("\n  DATA CLEANING COMPLETE ✓")
print("=" * 60)
```
